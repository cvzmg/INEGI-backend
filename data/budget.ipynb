```python
import pandas as pd
```

```python
data = pd.read_csv("./raw/budget/budget.csv")
data
```

```python
for i in range(1,len(data.columns)):
    index=data.columns[i]
    data[index+"_PORCENTAJE_MUN"]=data[index]/data[index].max()
data
```
