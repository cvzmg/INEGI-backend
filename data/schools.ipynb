```python
import pandas as pd
```

```python
data = pd.read_csv("./raw/education/escuelas en la zona metropolitana.xlsx - Totales de escuela.csv")
data
```

```python
data=data[["ent","alc","tipo_educ","Numero de escuelas"]]
data
```

```python
data=data.query('ent=="Ciudad de México"')
```

```python
data=data.drop(['ent'], axis=1)
```

```python
data_educación_basica = data[data["tipo_educ"].isin(["Básica", "Media Superior"])]
data_educación_basica= data_educación_basica.groupby("alc", as_index=False)["Numero de escuelas"].sum()
data_educación_basica.rename(columns={'alc': 'ALCALDIA', 'Numero de escuelas': 'NO_ESCUELAS_BASICAS'}, inplace=True)
data_educación_basica
```

```python
data_educación_superior = data[data["tipo_educ"].isin(["Superior"])]
data_educación_superior= data_educación_superior.groupby("alc", as_index=False)["Numero de escuelas"].sum()
data_educación_superior.rename(columns={'alc': 'NOS', 'Numero de escuelas': 'NO_ESCUELAS_SUPERIORES'}, inplace=True)
data_educación_superior
```

```python
numero_escuelas=pd.concat([data_educación_basica, data_educación_superior], axis=1)
numero_escuelas=numero_escuelas.drop('NOS',axis=1)
```

```python
new_row_df = pd.DataFrame({"ALCALDIA": ["Total de la entidad Ciudad de México"], "NO_ESCUELAS_BASICAS": [numero_escuelas["NO_ESCUELAS_BASICAS"].sum()],"NO_ESCUELAS_SUPERIORES": [numero_escuelas["NO_ESCUELAS_SUPERIORES"].sum()] })
numero_escuelas = pd.concat([new_row_df,numero_escuelas], ignore_index=False).reset_index().drop("index",axis=1)
numero_escuelas
```

```python
total_municipal=numero_escuelas.iloc[:,1:].sum(axis=1)
for i in range(1,len(numero_escuelas.columns)):
    index=numero_escuelas.columns[i]
    numero_escuelas[index+"_PORCENTAJE_MUN"]=numero_escuelas[index]/total_municipal
numero_escuelas
```

```python
numero_escuelas.to_csv("./output/schools.csv", index=False)
```
