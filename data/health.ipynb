```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("./raw/health/centros_de_salud.csv")
data=data.iloc[1:,:].reset_index().drop('index',axis=1)
data
```

```python
excluir = [
    "De consulta externa",
    "De hospitalización general",
    "De hospitalización especializada",
    "No especificado"
]

centros_salud = data[~data["Demarcación territorial"].isin(excluir)].reset_index(drop=True)
centros_salud
```

```python
centros_salud["NUMERO_CENTROS"] = centros_salud['NUMERO_CENTROS'].astype(int)
centros_salud
```

```python
centros_salud.loc[0, "NUMERO_CENTROS"] = centros_salud.iloc[1:,:]['NUMERO_CENTROS'].sum()
centros_salud
```

```python
centros_salud['NUMERO_CENTROS_PORCENTAJE']=centros_salud["NUMERO_CENTROS"]/centros_salud["NUMERO_CENTROS"][0]
centros_salud
```

```python
centros_salud.to_csv("./output/health_centers.csv", index=False)
```
