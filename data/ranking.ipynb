```python
import pandas as pd
from functools import reduce
```

```python
data_budget=pd.read_csv("./output/budget.csv")
data_housing=pd.read_csv("./output/housing_state.csv")
data_green_spaces=pd.read_csv("./output/green_spaces.csv")
data_health_centers=pd.read_csv("./output/health_centers.csv")
data_hotels=pd.read_csv("./output/hotels.csv")
data_restaurant=pd.read_csv("./output/restaurant.csv")
data_transportation=pd.read_csv("./output/transportation.csv")
data_security=pd.read_csv("./output/security_everyone.csv")
```

```python
data_budget=data_budget.sort_values(by="Alcaldia").rename(columns={"Alcaldia":"ALCALDIA"})
data_housing=data_housing.iloc[1:,:].sort_values(by="NOM_MUN").rename(columns={"NOM_MUN":"ALCALDIA"})
data_green_spaces=data_green_spaces.iloc[1:,:].sort_values(by="Alcaldía").rename(columns={"Alcaldía":"ALCALDIA"})
data_health_centers=data_health_centers.iloc[1:,:].sort_values(by="Demarcación territorial").rename(columns={"Demarcación territorial":"ALCALDIA"})
data_restaurant=data_restaurant.sort_values(by="alcaldia").rename(columns={"alcaldia":"ALCALDIA"})
data_transportation=data_transportation.iloc[1:,:].sort_values(by="Alcaldias").rename(columns={"Alcaldias":"ALCALDIA"})
data_security=data_security.sort_values(by="NOM_MUN").rename(columns={"NOM_MUN":"ALCALDIA"})
```

```python
merged_data=data_budget
merge=[data_housing,data_green_spaces,data_health_centers,data_restaurant,data_transportation,data_security]
for i in range(0,len(merge)):
    merged_data=merged_data.merge(merge[i], on="ALCALDIA")
merged_data
```

```python
merged_data.columns
```

```python
merged_data_percentage=merged_data[['ALCALDIA', 'venta_PORCENTAJE_MUN','Renta_PORCENTAJE_MUN','BUEN_EDO','Superficie (m²)_PORCENTAJE',
                                   'NUMERO_CENTROS_PORCENTAJE','tipo','Lineas de metro_PORCENTAJE_MUN', 'Lineas de metrobus_PORCENTAJE_MUN',
                                   'Estaciones Ecobici_PORCENTAJE_MUN', 'LineasRTP_PORCENTAJE_MUN',"normalized"]]
```

```python
merged_data_percentage['PUBLIC_TRANSPORATION']=merged_data_percentage.iloc[:,7:11].mean(axis=1)
merged_data_percentage
```

```python
merged_data_percentage['HOUSING_PRICE']=merged_data_percentage.iloc[:,1:3].mean(axis=1)
merged_data_percentage
```

```python
percentage_alcaldia=merged_data_percentage[["Superficie (m²)_PORCENTAJE","BUEN_EDO","tipo","PUBLIC_TRANSPORATION","HOUSING_PRICE","normalized"]].mean(axis=1)
```

```python
percentage_alcaldia
```

```python
filtered_alcaldia=pd.concat([merged_data_percentage["ALCALDIA"],percentage_alcaldia], axis=1)
filtered_alcaldia=filtered_alcaldia.rename(columns={0:"PORCENTAJE"})
filtered_alcaldia
```

```python
alcaldia_ranking= filtered_alcaldia.sort_values(by="PORCENTAJE", ascending=False).reset_index().drop("index", axis=1)
alcaldia_ranking
```

```python
alcaldia_ranking.to_csv("./output/ranking_quality.csv", index=False)
```
