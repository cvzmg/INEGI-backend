```python
import pandas as pd
import numpy as np
```

### Hoteles

```python
data = pd.read_csv("./raw/cultural_centers/establecimientos-de-hospedaje-2021.csv",encoding='latin-1')
data=data.iloc[:,3:6].drop(["CP"], axis=1)
```

```python
data
```

```python
data['CATEGORÍA'] = data['CATEGORÍA'].str.replace(' Estrellas', '', regex=False)
data['CATEGORÍA'] = data['CATEGORÍA'].str.replace(' estrellas', '', regex=False)
data['CATEGORÍA'] = data['CATEGORÍA'].str.replace(' Estrella', '', regex=False)
data['CATEGORÍA'] = data['CATEGORÍA'].replace('Sin Clasificar', np.nan)
```

```python
data["CATEGORÍA"]=pd.to_numeric(data["CATEGORÍA"], errors="coerce")
```

```python
data=data.groupby("ALCALDÍA")["CATEGORÍA"].mean()
data
```

```python
df_to_save = data.reset_index()

df_to_save.columns = ['ALCALDÍA', 'CATEGORIA_PROMEDIO']
```

```python
df_to_save.to_csv("./output/hotels.csv", index=False)
```

### Servicios turisticos

```python
data = pd.read_csv("./raw/cultural_centers/prestadores-de-servicios-turisticos-de-la-ciudad-de-mexico.csv")
data
```

```python
data_tipo=data["tipo"]
data_alcaldia=data["alcaldia"]
```

```python
data_restaurantes=pd.concat([data_alcaldia,data_tipo], axis=1)
data_restaurantes
```

```python
data_restaurantes=data_restaurantes.query('tipo=="Restaurante"').replace('Restaurante', int(1), regex=False)
```

```python
data_restaurantes=data_restaurantes.groupby("alcaldia")["tipo"].sum()
```

```python
data_restaurantes
```

```python
df_to_save = data_restaurantes.reset_index()

df_to_save.columns = ['alcaldia', 'tipo']
df_to_save
```

```python
df_to_save["tipo"]=df_to_save["tipo"]/df_to_save["tipo"].sum()
```

```python
df_to_save
```

```python
df_to_save.to_csv("./output/restaurant.csv", index=False)
```
