```python
import pandas as pd
```

### History and Culture


```python
data = pd.read_csv("./raw/life_style/historic.csv")
data_transposed = data.T
data_transposed=data_transposed.iloc[1:,:]
data_transposed["HISTORIC_SUM"]=data_transposed.sum(axis=1)
data_transposed["HISTORIC_PERCENTAGE"]=data_transposed["HISTORIC_SUM"]/data_transposed["HISTORIC_SUM"].sum()
df = data_transposed["HISTORIC_PERCENTAGE"].reset_index().rename(columns={'index': 'ALCALDIA'})
df=df.sort_values(by="ALCALDIA")
df.to_csv("./output/culture.csv", index=False)
```
