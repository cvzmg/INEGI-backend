```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./raw/tenement/conjunto_de_datos_iter_09CSV20.csv")
```

```python
data= df[[ "NOM_MUN","NOM_LOC","POBTOT","POBFEM","POBMAS","P_18YMAS","P_18YMAS_F","P_18YMAS_M","P_60YMAS","P_60YMAS_F","P_60YMAS_M","P_18A24","P_18A24_F","P_18A24_M","P_25A29","P_25A29_F","P_25A29_M","PCON_DISC","P12YM_CASA","P12YM_SOLT","P12YM_SEPA","PCATOLICA","PRO_CRIEVA","POTRAS_REL","PSIN_RELIG","PROM_OCUP","VPH_PISODT","VPH_PISOTI","VPH_C_ELEC","VPH_S_ELEC","VPH_AEASP","VPH_AGUAFV","VPH_DRENAJ","VPH_NODREN","VPH_SNBIEN","VPH_REFRI","VPH_LAVAD","VPH_HMICRO","VPH_AUTOM","VPH_MOTO","VPH_BICI","VPH_RADIO","VPH_TV","VPH_PC","VPH_TELEF","VPH_CEL","VPH_INTER","VPH_STVP","VPH_SPMVPI","VPH_CVJ"]]
data
```

```python
data=data.query('NOM_LOC=="Total del Municipio" or NOM_LOC=="Total de la Entidad"').reset_index()
data
```

```python
data=data.drop(["index","NOM_LOC"], axis=1)
data
```

```python
dr=data.iloc[:,0]
dr
```

```python
dh=data.iloc[:,1:].apply(pd.to_numeric, errors="coerce")
dh
```

```python
data=pd.concat([dr,dh],axis=1)
data
```

```python
nombres_alcaldias=data.iloc[:,0]
nombres_alcaldias
```

### Poblacion edades

```python
poblacion_y_edad=data.iloc[:,:16]
poblacion_y_edad
```

```python
for r in range(1,15+1,3):
    for i in range(0,2+1):
        index=poblacion_y_edad.columns[r+i]
        poblacion_y_edad[index+"_PORCENTAJE_MUN"]=poblacion_y_edad[index]/poblacion_y_edad[poblacion_y_edad.columns[r]]
poblacion_y_edad
```

```python
poblacion_y_edad.to_csv("./output/ages.csv", index=False)
```

### Poblacion casada

```python
poblacion_casada=data.iloc[:,17:20]
poblacion_casada=pd.concat([nombres_alcaldias,poblacion_casada], axis=1)
poblacion_casada
```

```python
total_municipal=poblacion_casada.iloc[:,1:4].sum(axis=1)
for i in range(1,len(poblacion_casada.columns)):
    index=poblacion_casada.columns[i]
    poblacion_casada[index+"_PORCENTAJE_MUN"]=poblacion_casada[index]/total_municipal
poblacion_casada
```

```python
poblacion_casada.to_csv("./output/married.csv", index=False)
```

### Poblacion religiosa

```python
poblacion_religiosa=data.iloc[:,20:24]
poblacion_religiosa=pd.concat([nombres_alcaldias,poblacion_religiosa], axis=1)
poblacion_religiosa
```

```python
total_municipal=poblacion_religiosa.iloc[:,1:5].sum(axis=1)
for i in range(1,len(poblacion_religiosa.columns)):
    index=poblacion_religiosa.columns[i]
    poblacion_religiosa[index+"_PORCENTAJE_MUN"]=poblacion_religiosa[index]/total_municipal
poblacion_religiosa
```

```python
poblacion_religiosa.to_csv("./output/religion.csv", index=False)
```

### Estado de vivienda, bienes y proporción de personas por vivienda

```python
estado_vivienda=data.iloc[:,24:]
estado_vivienda=pd.concat([nombres_alcaldias,estado_vivienda],axis=1)
estado_vivienda
```

```python
buen_edo_suelo=estado_vivienda['VPH_PISODT']/(estado_vivienda['VPH_PISODT']+estado_vivienda['VPH_PISOTI'])
buen_edo_electricidad=estado_vivienda['VPH_C_ELEC']/(estado_vivienda['VPH_C_ELEC']+estado_vivienda['VPH_S_ELEC'])
buen_edo_agua=estado_vivienda['VPH_AEASP']/(estado_vivienda['VPH_AEASP']+estado_vivienda['VPH_AGUAFV'])
buen_edo_drenaje=estado_vivienda['VPH_DRENAJ']/(estado_vivienda['VPH_DRENAJ']+estado_vivienda['VPH_NODREN'])
buen_edo_bienes=estado_vivienda[['VPH_REFRI', 'VPH_LAVAD', 'VPH_HMICRO', 'VPH_AUTOM','VPH_MOTO', 'VPH_BICI', 'VPH_RADIO', 'VPH_TV', 'VPH_PC', 'VPH_TELEF','VPH_CEL', 'VPH_INTER', 'VPH_STVP', 'VPH_SPMVPI', 'VPH_CVJ']].sum(axis=1)/(estado_vivienda['VPH_SNBIEN']+estado_vivienda[['VPH_REFRI', 'VPH_LAVAD', 'VPH_HMICRO', 'VPH_AUTOM','VPH_MOTO', 'VPH_BICI', 'VPH_RADIO', 'VPH_TV', 'VPH_PC', 'VPH_TELEF','VPH_CEL', 'VPH_INTER', 'VPH_STVP', 'VPH_SPMVPI', 'VPH_CVJ']].sum(axis=1))

buen_edo_general=((buen_edo_suelo+buen_edo_electricidad+buen_edo_agua+buen_edo_drenaje+buen_edo_bienes)/5)
```

```python
estado_vivienda["BUEN_EDO"]=buen_edo_general
```
