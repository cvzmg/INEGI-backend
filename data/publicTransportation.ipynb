```python
import pandas as pd
```

```python
data = pd.read_csv("./raw/transportation/transporte_publico - Hoja 1.csv")
data
```

```python
new_row_df = pd.DataFrame({"Alcaldias": ["Total de la entidad Ciudad de México"], "Lineas de metro": [data["Lineas de metro"].sum()],"Lineas de metrobus": [data["Lineas de metrobus"].sum()],"Estaciones Ecobici": [data["Estaciones Ecobici"].sum()],"LineasRTP": [data["LineasRTP"].sum()]})
numero_transporte = pd.concat([new_row_df,data], ignore_index=False).reset_index().drop("index",axis=1)
numero_transporte
```

```python
for i in range(1,len(numero_transporte.columns)):
    index=numero_transporte.columns[i]
    numero_transporte[index+"_PORCENTAJE_MUN"]=numero_transporte[index]/numero_transporte[index][0]
numero_transporte
```

```python
numero_transporte.to_csv("./output/transportation.csv", index=False)
```
